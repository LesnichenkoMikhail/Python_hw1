Напишите такой класс Animal, чтобы при вызове его классового метода get_highest() возвращался рост самого высокого из экземпляров

```python
class Animal():
  animals = []
  def __init__(self, name, height):
    self.name = name 
    self.height = height
    Animal.animals.append(self)
  

  def speak(self):
    print("...")

  def __eq__(self, other):
    return type(self) == type(other)

  @classmethod
  def get_highest(cls):
    return max(animal.height for animal in cls.animals) 
```

Это все здорово, но как-то неправильно, что класс Animal имеет метод speak, а классы-потомки не определяют свою реализацию. Давайте сделаем так, чтобы все виды животных были обязаны иметь свою реализацию метода speak, но просто animal'ов уже нельзя было создать

```python

from abc import ABC, abstractmethod


class Animal(ABC):
    animals = []

    def __init__(self, name, height):
        self.name = name
        self.height = height
        Animal.animals.append(self)

    @abstractmethod
    def speak(self):
        pass

    @classmethod
    def get_highest(cls):
        return max(animal.height for animal in cls.animals)

    def __eq__(self, other):
        return type(self) == type(other)


class Lion(Animal):

    def speak(self):
        print('ROOOOAR!!!!!')


class Lemur(Animal):

    def speak(self):
        print("Lemur sounds uwu")


class Giraffe(Animal):

    def speak(self):
        print('Giraffe sounds (Звукии мнительного Мелмана)')
        
```

```python
from abc import ABC, abstractmethod
from collections import Counter


class UniversalFormat:

    def __init__(self, entries, language=None):
        self.entries = entries
        self.language = language

    def get_language(self):
        if self.language is not None:
            return self.language
        return "unknown"


class Format(ABC):

    @abstractmethod
    def parse(self, text):
        pass

    @abstractmethod
    def serialize(self, data):
        pass


def stats(files):
    languages = []

    for format_, text in files:
        data = format_.parse(text)
        languages.append(data.get_language())

    return Counter(languages)
```

```python
import csv
import io


class CsvFormat(Format):

    def parse(self, text):
        text = text.strip()

        reader = csv.DictReader(
            io.StringIO(text),
            skipinitialspace=True
        )

        entries = []

        for row in reader:
            entries.append({
                "id": row["Title"],
                "source": row["en"],
                "target": row["de-AT"]
            })

        return UniversalFormat(
            entries,
            language="de-AT"
        )

    def serialize(self, data):
        output = io.StringIO()

        writer = csv.writer(output)

        writer.writerow(["Title", "Description", "en", "de-AT"])

        for entry in data.entries:
            writer.writerow([
                entry["id"],
                "",
                entry["source"],
                entry["target"]
            ])

        return output.getvalue()
```

```python
import re


def detect_language(text):
    text = text.lower()

    words = set(
        re.findall(r"\b[\wÀ-ÿ'-]+\b", text)
    )

    french_words = {
        "tu", "ça", "une", "avec", "pour",
        "suivants", "extrait", "traductions",
        "commence", "appuie", "sauvegarder"
    }

    german_words = {
        "hurra", "bist", "der", "die", "das",
        "eine", "textpassage", "übersetzungen",
        "deinen", "deine", "nächste", "gespeichert"
    }

    english_words = {
        "welcome", "you're", "the", "day", "better",
        "following", "text", "translations",
        "start", "your", "save", "next", "made", "with"
    }

    counts = {
        "fr": len(words & french_words),
        "de": len(words & german_words),
        "en": len(words & english_words)
    }

    max_count = max(counts.values())

    if max_count == 0:
        return "unknown"

    languages = [
        language
        for language, count in counts.items()
        if count == max_count
    ]

    if len(languages) > 1:
        return "unknown"

    return languages[0]
```

```python
import xml.etree.ElementTree as ET
import re


class ReswFormat(Format):

    def parse(self, text):
        text = text.strip()
        root = ET.fromstring(text)

        entries = []

        for data in root.findall("data"):
            value = data.findtext("value", default="")

            entries.append({
                "id": data.attrib.get("name", ""),
                "source": value,
                "target": value
            })

        language = detect_language(
            " ".join(
                entry["source"]
                for entry in entries
            )
        )

        return UniversalFormat(
            entries,
            language=language
        )

    def serialize(self, data):
        root = ET.Element("root")

        for entry in data.entries:
            data_element = ET.SubElement(
                root,
                "data",
                {"name": entry["id"]}
            )

            value = ET.SubElement(
                data_element,
                "value"
            )

            value.text = entry["target"]

        return ET.tostring(
            root,
            encoding="unicode"
        )
```

```python
class PhpFormat(Format):

    def parse(self, text):
        entries = []

        pattern = r'define\("([^"]+)",\s*"([^"]*)"\);'

        for match in re.finditer(pattern, text):
            key = match.group(1)
            value = match.group(2)

            entries.append({
                "id": key,
                "source": value,
                "target": value
            })

        language = detect_language(
            " ".join(
                entry["source"]
                for entry in entries
            )
        )

        return UniversalFormat(
            entries,
            language=language
        )

    def serialize(self, data):
        result = ["<?php"]

        for entry in data.entries:
            result.append(
                f'  define("{entry["id"]}", "{entry["target"]}");'
            )

        result.append("?>")

        return "\n".join(result)
```

```python
class XliffFormat(Format):

    def parse(self, text):
        root = ET.fromstring(text.strip())

        namespace = {
            "x": "urn:oasis:names:tc:xliff:document:1.2"
        }

        file_element = root.find("x:file", namespace)

        language = file_element.attrib.get(
            "target-language"
        )

        entries = []

        for unit in root.findall(
            ".//x:trans-unit",
            namespace
        ):
            source = unit.find("x:source", namespace)
            target = unit.find("x:target", namespace)

            entries.append({
                "id": unit.attrib.get(
                    "resname",
                    unit.attrib.get("id")
                ),
                "source": source.text if source is not None else "",
                "target": target.text if target is not None else ""
            })

        return UniversalFormat(
            entries,
            language=language
        )

    def serialize(self, data):
        root = ET.Element(
            "xliff",
            {
                "version": "1.2",
                "xmlns": "urn:oasis:names:tc:xliff:document:1.2"
            }
        )

        file_element = ET.SubElement(
            root,
            "file",
            {
                "source-language": "en",
                "target-language": data.language or "unknown",
                "datatype": "plaintext"
            }
        )

        body = ET.SubElement(file_element, "body")

        for index, entry in enumerate(data.entries, 1):
            unit = ET.SubElement(
                body,
                "trans-unit",
                {
                    "id": str(index),
                    "resname": entry["id"]
                }
            )

            source = ET.SubElement(unit, "source")
            source.text = entry["source"]

            target = ET.SubElement(unit, "target")
            target.text = entry["target"]

        return ET.tostring(
            root,
            encoding="unicode"
        )
```

```python
class PoFormat(Format):

    def parse(self, text):
        entries = []

        msgid = None
        msgstr = None

        for line in text.splitlines():

            if line.startswith("msgid "):
                msgid = line[6:].strip().strip('"')

            elif line.startswith("msgstr "):
                msgstr = line[7:].strip().strip('"')

                if msgid is not None and msgid != "":
                    entries.append({
                        "id": msgid,
                        "source": msgid,
                        "target": msgstr
                    })

                msgid = None
                msgstr = None

        language = detect_language(
            " ".join(
                entry["target"]
                for entry in entries
            )
        )

        return UniversalFormat(
            entries,
            language=language
        )

    def serialize(self, data):
        result = []

        for entry in data.entries:
            result.append(
                f'msgid "{entry["source"]}"'
            )
            result.append(
                f'msgstr "{entry["target"]}"'
            )
            result.append("")

        return "\n".join(result)
```

```python
class TsFormat(Format):

    def parse(self, text):
        text = text.strip()
        root = ET.fromstring(text)

        language = root.attrib.get("language")

        entries = []

        for message in root.findall(".//message"):
            source = message.find("source")
            translation = message.find("translation")

            entries.append({
                "id": message.attrib.get("id", ""),
                "source": source.text if source is not None else "",
                "target": translation.text if translation is not None else ""
            })

        return UniversalFormat(
            entries,
            language=language
        )

    def serialize(self, data):
        root = ET.Element(
            "TS",
            {
                "version": "2.1",
                "language": data.language or "unknown"
            }
        )

        context = ET.SubElement(root, "context")

        for entry in data.entries:
            message = ET.SubElement(
                context,
                "message",
                {"id": entry["id"]}
            )

            source = ET.SubElement(message, "source")
            source.text = entry["source"]

            translation = ET.SubElement(message, "translation")
            translation.text = entry["target"]

        return ET.tostring(root, encoding="unicode")
```

Как именно изменится ваша модель классов для того, чтобы поддержать новые форматы?

Для новых форматов я добавляю отдельный класс, который наследуется от Format. В нём прописываю два метода: parse() для чтения файла и serialize() для сохранения файла.

Все файлы сначала переводятся в общий формат UniversalFormat, поэтому их можно легко переводить из одного формата в другой.

Например: PO → UniversalFormat → CSV.

Поэтому, чтобы добавить новый формат, достаточно создать новый класс и реализовать в нём эти два метода.
